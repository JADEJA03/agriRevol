```python
from ultralytics import YOLO


model = YOLO('yolov8n.pt')  # Load the lightweight model

# Train model
model.train(data='/content/drive/MyDrive/data_try_2/data.yaml', epochs=50, imgsz=640, batch=16, device='cuda')
```

```python

metrics = model.val()
```
