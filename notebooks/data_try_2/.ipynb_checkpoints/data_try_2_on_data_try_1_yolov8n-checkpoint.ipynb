```python
from ultralytics import YOLO

# 1. Load your previously trained model from data_try_1
model = YOLO('/content/drive/MyDrive/yolov8n_data_try_1/yolo_best.pt')
```

```python
model.train(
    data='/content/drive/MyDrive/data_try_2/data.yaml',  # path to new data config
    epochs=50,
    imgsz=640,
    batch=16,
    device='cuda',  # or 'cpu' if no GPU
    freeze=10  # freeze first 10 layers
)
```

```python

metrics = model.val()
```
