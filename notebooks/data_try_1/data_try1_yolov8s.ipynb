```python
from ultralytics import YOLO

# Load model
model = YOLO('yolov8s.pt')

# Train model
model.train(data='/content/drive/MyDrive/data_try1/data.yaml', epochs=50, imgsz=640, batch=16, device='cuda')
```

```python
metrics = model.val()
```

```python
results = model('/content/drive/MyDrive/data_try1/images/test/DSC00844.jpg')

# Access the first result in the list
results[0].show()  # Show output
results[0].save('/content/drive/MyDrive/results/')  # Save results
```
